```python
from statsbombpy import sb
import warnings
import pandas as pd

from datadownloader import DataDownloader
from datasetcreator import columns_to_keep, DataSetCreator
```

# Download all games

```python
# Call statsbombpy API to get all free competitions
free_comps = sb.competitions()
free_comps.head()
```

```python
competition_ids = [
    11, # LaLiga
    2, # Premier League
    12, # Serie A
    7, # Ligue 1
    9, # Bundesliga
]
```

```python
warnings.filterwarnings("ignore", message="credentials were not supplied")
for competition_id in competition_ids:
    free_seasons = free_comps[free_comps['competition_id'] == competition_id]
    downloader = DataDownloader(
        free_seasons, 
        directory='..\\..\\data-storage\\raw\\top_5_leagues',
        )
    downloader.download_store_seasons()
```

```python
# Select UEFA Euros 2020
free_seasons = free_comps[free_comps['season_id'] == 43]
downloader = DataDownloader(
    free_seasons, 
    directory='..\\..\\data-storage\\raw\\euros_2020',
    )
downloader.download_store_seasons()
```

```python
df_events = pd.read_parquet(
    '../../data-storage/preprocessed/top_5_leagues.parquet', 
    engine='fastparquet',
    )

df_events['shot'] = ~df_events['shot_outcome'].isna()
df_events['goal'] = df_events['shot_outcome'] == 'Goal'

df_events.drop(columns=['id', 'shot_outcome', 'possession'], inplace=True)
df_size_bytes = df_events.memory_usage(deep=True).sum()
df_size_mb = df_size_bytes / (1024 ** 2)  # Convert to MB
print(f"DataFrame size: {df_size_mb:.2f} MB")
print(f"DataFrame shape: {df_events.shape}")
```
