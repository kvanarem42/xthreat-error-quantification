```python
import pandas as pd
import numpy as np
```

```python
df_old = pd.read_parquet('../../data-storage/preprocessed/ligue1_1516.parquet')
df_new = pd.read_parquet('../../data-storage/preprocessed/ligue1_2015_2016_test.parquet')
```

```python
df_old.describe()
```

```python
df_new.describe()
```

```python
for col in df_old.columns:
    if col not in df_new.columns:
        print(f"Column '{col}' is missing in the new DataFrame.")
    else:
        old_dtype = df_old[col].dtype
        new_dtype = df_new[col].dtype
        if old_dtype != new_dtype:
            print(f"Column '{col}' has different data types: old={old_dtype}, new={new_dtype}.")
```

```python
for col in df_old.columns:
    if col not in df_new.columns:
        print(f"Column '{col}' is missing in the new DataFrame.")
    else:
        equal = np.all(df_old[col].values == df_new[col].values)
        if not equal:
            print(f"Column '{col}' has different values.")
```
